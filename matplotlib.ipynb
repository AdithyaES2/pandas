```python
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=[40,15,22,10]
plt.plot(x,y) #to print or draw the line chart
plt.show() # to display the value
```

```python
month=["jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
plt.plot(month,sales)
plt.show()
```

```python
# plot chart
month=["jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
profit=[5,6,8,12,15]
plt.plot(month,sales,label="sales")# LABEL: NAME THE LABEL
plt.plot(month,profit,label="profit")
plt.legend()# legend : to display  the label
plt.xlabel("MONTHS") # to mark x axis
plt.ylabel("SALES") # to mark y axis
plt.title("MONTHLY SALES AND PROFIT")
plt.grid(True) # to draw grid
plt.plot(
     month,
     sales,
     marker="*", #o:circle ,s:square
     color="red", # to change the color
     linestyle="dashed" # linestyle
)

plt.show()# show: to display the value
```

BAR CHART

```python
#  bar chart
month=["jan","Feb","Mar","Apr","May"]
sales=[20,30,25,40,50]
profit=[5,6,8,12,15]
plt.xlabel("MONTHS") # to mark x axis
plt.ylabel("SALES") # to mark y axis
plt.title("MONTHLY SALES AND PROFIT")
plt.grid(True) # to draw grid
plt.bar(
    month,
    sales,
    color="red")

plt.show()# show: to display the value
```

SCATTER

```python
#scatter
study_hours=[1,2,3,4,5,6]
marks=[40,45,55,65,75,85]
plt.scatter(study_hours,marks) # scatter:dot chart
plt.xlabel("study hours")
plt.ylabel("marks")
plt.title("study hours vs marks")
plt.show()
```

HISTOGRAM

```python

marks=[40,45,55,65,75,85,90,95,35,76,34,87,76]
plt.hist(marks)
plt.title("marks distribution")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.show()
```

PIE CHART

```python
categories=["Food","Travel","Shopping","Others"]
amount=[4000,2000,3000,1000]
plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")
plt.show()
```

```python
fig,ax=plt.subplots(2,1) #fig:whole graph,ax:individual graph,(2,1):row & column
ax[0].plot([1,2,3,4],[10,20,30,40])
ax[1].bar([1,2,3,4],[40,30,20,10])
plt.show()

```
