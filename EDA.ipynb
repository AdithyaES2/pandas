```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('titanic.csv')
df
```

```python
# to find null values , sum()->to find the no. of errors in each column
df.isnull().sum()
df["Embarked"]=df["Embarked"].fillna(
    df["Embarked"].mode()[0])
print(df.isnull().sum())
print("Before:",df.Age.shape[0])
df=df.drop_duplicates()

print("after:",df.shape[0])
df=df.rename(columns={
    "PassengerId":"PassengerId",
    "Survived":"Survived"
})

df["Sex"]=df["Sex"].map({
    "male":0,
    "female":1
})



```

```python
sns.histplot(data=df, x="Age", bins=20)
plt.title("Age Distribution of titanic passenger")
plt.xlabel("Age")
plt.ylabel("Number of passengers")
plt.show()

sns.countplot(data=df, x="Sex", hue="2urvived")
plt.title("Sex Distribution of titanic passenger")
plt.xlabel("Sex")
plt.ylabel("Number of passengers")
plt.show()

sns.countplot(data=df, x="Pclass", hue="2urvived")
plt.title("Pclass Distribution of titanic passenger")
plt.xlabel("Pclass")
plt.ylabel("Number of passengers")
plt.show()
```
