```python
#data frame
import numpy as np
import pandas as pd

data={
    "Name":["Adithya","Fidha","Rihana","Shifa"],
    "Age":[20,17,19,18],
    "Marks":[85,72,np.nan,90],
    "Attendance":[95,80,75,np.nan],
    "city":["Kochi","Delhi","Mumbai","Chennai"]
}
df=pd.DataFrame(data)
print(df)


print(df.head(2))# first2 row & column
print(df.tail(1))#last1 row & column
print(df.shape)#to find no.of rows and columns
print(df.columns)
print(df.dtypes)#to find data type
print(df["Name"])#to find a specific data
print(df.iloc[:3])# skip 3rd data
print(df.loc[:3])# include 3rd data

print(df[df["Marks"]>75])
print(df[df["city"]=="Mumbai"])
print(df[(df["Marks"]>75) & (df["Attendance"]>80)])
print("\nMarks + 5:")
print(df["Marks"]+5)
print("\nMarks *2:")
print(df["Marks"]*2)
print("\nAverage Marks:")
print(df["Marks"].mean())
print("\nMaximum Marks:")
print(df["Marks"].max())
print("\nMinimum Marks:")
print(df["Marks"].min())
print("\nTotal Marks:")
print(df["Marks"].sum())
print(df.sort_values("Marks"))
print(df.sort_values("Marks",ascending=False))
print(df.rename(
    columns={
    "marks":"score","Attendance":"attendence_percentage"
    }
))
print(df.isna())
print(df.dropna())
print("\n",df.fillna(0))

df_for=df.copy()
df_for=df_for.ffill()
print("\n",df_for)

df_for=df.copy()
df_for=df_for.bfill()
print("\n",df_for)
```
