```python
#loc
datas=[1,2,3,4,5]
s=pd.Series(datas,index=["a","b","c","d","e"])
print(s.loc["d"])


#iloc

print(s.iloc[0])
```

```python
import pandas as pd
sales=pd.Series([1200,850,1500,950,2000],
                index=["Monday","Tuesday","Wednesday","Thursday","Friday"])

print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])



```

```python
#remove duplicates
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
```

```python
# count of unique functions
import pandas as pd
data=[20,None,30,None,50,60,None,80]
ar=pd.Series(data)
print(ar)
print(ar.isnull())
print(ar.notnull())
print(ar.nunique())
print(ar.value_counts())
```
